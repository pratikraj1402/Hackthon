```python
import os
import torch
from datasets import Dataset
from transformers import (
    AutoModelForSeq2SeqLM,
    AutoTokenizer,
    DataCollatorForSeq2Seq,
    Seq2SeqTrainer,
    Seq2SeqTrainingArguments,
)

# Optional: suppress the Hugging Face hub rate-limit warning
# os.environ["HF_TOKEN"] = "your_hf_token_here"

# 1. Dataset Preparation
training_data = {
    "input_text": [
        "Patient: John Doe, 45, Male. Complaint: Severe chest pain for 3 days. Severity: Moderate. History: Hypertension. OCR Diagnoses: Angina. OCR Meds: Tab Aspirin 75mg. OCR Tests: BP: 140/90 mmHg.",
        "Patient: Mary Smith, 32, Female. Complaint: High fever and cough for 5 days. Severity: Severe. History: Asthma. OCR Diagnoses: Acute Bronchitis. OCR Meds: Syr Amoxicillin 500mg. OCR Tests: SpO2: 95%.",
        "Patient: Alex R, 60, Male. Complaint: Joint stiffness for 2 weeks. Severity: Mild. History: Diabetes. OCR Diagnoses: Osteoarthritis. OCR Meds: Tab Paracetamol 500mg. OCR Tests: HbA1c: 7.2%.",
    ],
    "target_text": [
        "CLINICAL SUMMARY:\n45 Y/O Male presented with moderate chest pain (3 days) with underlying hypertension. OCR indicates Angina with BP 140/90 mmHg. Currently taking Aspirin 75mg. Requires cardiac follow-up.",
        "CLINICAL SUMMARY:\n32 Y/O Female presented with severe fever and cough (5 days) with underlying Asthma. OCR indicates Acute Bronchitis with SpO2 at 95%. Prescribed Amoxicillin 500mg. Monitor respiratory status.",
        "CLINICAL SUMMARY:\n60 Y/O Male presenting with mild joint stiffness (2 weeks). History of Diabetes (HbA1c 7.2%). Diagnosis: Osteoarthritis. Managed on Paracetamol 500mg.",
    ],
}

dataset = Dataset.from_dict(training_data)

# 2. Load Tokenizer & Model
MODEL_NAME = "google/flan-t5-base"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)


# 3. Tokenize Inputs and Targets
def preprocess(examples):
    inputs = [
        "summarize clinical data: " + text for text in examples["input_text"]
    ]
    model_inputs = tokenizer(inputs, max_length=512, truncation=True)
    labels = tokenizer(
        text_target=examples["target_text"], max_length=150, truncation=True
    )

    # Replace pad tokens with -100 so the cross-entropy loss ignores them
    labels["input_ids"] = [
        [
            (token if token != tokenizer.pad_token_id else -100)
            for token in label
        ]
        for label in labels["input_ids"]
    ]

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs


tokenized_dataset = dataset.map(preprocess, batched=True)

# Use dynamic padding during batching for faster training
data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model,
    label_pad_token_id=-100,
    pad_to_multiple_of=8 if torch.cuda.is_available() else None,
)

# 4. Training Arguments
training_args = Seq2SeqTrainingArguments(
    output_dir="./doctor_summary_model",
    per_device_train_batch_size=2,
    learning_rate=3e-4,
    num_train_epochs=5,
    save_total_limit=1,
    fp16=torch.cuda.is_available(),
    logging_steps=1,
)

# 5. Trainer Initialization & Execution
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    data_collator=data_collator,
    processing_class=tokenizer,  # Replaced deprecated 'tokenizer' argument
)

print("Starting model fine-tuning...")
trainer.train()

# 6. Save Model Locally
model.save_pretrained("./final_clinical_model")
tokenizer.save_pretrained("./final_clinical_model")
print("Model fine-tuning complete and saved!")
```
